```python
import re

with open("sample.txt", "r") as file:
    text = file.read()

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', text)

with open("emails.txt", "w") as file:
    for email in emails:
        file.write(email + "\n")

print("Extracted Email Addresses:")
for email in emails:
    print(email)
```
